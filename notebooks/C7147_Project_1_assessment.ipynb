## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Validate an individual read against a reference sequence.

    Parameters
    ----------
    sequence : str
      the individual sequences in list of reads that need to be validates

    reference : str
      a non-empty reference sequence with only uppercase A G T C DNA nucleotides

    Returns
    -------
    dict

      A dictionary with keys:
        - "sequence" - str - forced uppercase individual sequneces
        - "valid" - bool - says wether the individual read's sequence length
                           matches the reference's lenght as well as wether its
                           letter are uppercase
        - "errors" - list containing str - list that has invalid reads and/or
                                           invalid characters, or and empty
                                           space ([]) if valid

         """
    #force uppercase letters
    uppercase_sequence = sequence.upper()

    #creating a list named arrors to collect errors
    errors = []

    #checking the length
    if len(uppercase_sequence) != len(reference):
        errors.append("wrong_length")

    #check for valid characters
    valid_bases = {"A", "C", "G", "T"}
    if not set(uppercase_sequence).issubset(valid_bases):
        errors.append("invalid_characters")

    #determining validity
    is_valid = len(errors) == 0

    return {"sequence": uppercase_sequence,
            "valid": is_valid,
            "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Find nucleotide base mismatches in valid individual sequences when
       compared to reference sequence.

       Parameters
       ----------

       sequence : str
          a valid individual read sequence

       reference : str
          a reference sequence for individual reads sequences

       Returns
       -------

       list of dict
          A list of dictionaries with order of:
            - "position" - int - position of the nucleotide mismatch in a range
                                 from 1 to 200 in an ascnding order
            - "reference_base" - str - nuclotide in the reference sequence
            -"read_base" - str - nucleotide in the read sequence
        """
    # Write your code here.
    #addidng an empty list to collect mismatches in sequnces when comapared ro reference
    mutations = []
    #for loop to pair bases in seq and ref and assigns them to position
    for position, (read_base, reference_base) in enumerate(zip(sequence, reference), start = 1):

      #checking for mismatch in every iteration and gives the information of the mismatch
      if read_base != reference_base:
        mutations.append({"position": position,
                          "reference_base": reference_base,
                          "read_base": read_base})
    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]

```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """ Validating and finding the mutations by identifying the mismatches in
        nucleotide bases between individual read sequences and reference sequence.

        The function designed not to interfere with original data.


        Parameters
        ----------
        reads : list of dict
          list of dictionaries, each of which contains keys:
            - "read_id" - str - id for individual reads
            - "sample" - str - sample name
            - "group" - str - treatment or control group
            - "sequence" - str - normalised individual nucleotide sequences

        reference : str
          a reference sequence for each individual reads

        Returns
        -------
        list of dict
          a list of dictionaries created. it contains the keys that has reads
          (from parameters above) and additional ones such as :
             "status" - str - identifies the validity of individual sequence
             "errors" - list of str - list of errors arised during validation,
                                      if any
             "mutations" - list of dict - list of mismatches identified with the
                                          postion in the sequence and the
                                          nuclotide mismatch (keys for this dict
                                          are specified in Step 2 docstring)
    """
    # Write your code here.
    #creating an empty list for analysed reads
    analyse_list = []
    #for loopto go through each read in dict and gives it to validate_read
    for read in reads:
      validating = validate_read(read["sequence"], reference)
      valid = validating["valid"]
      status = "valid" if valid else "invalid"
      #makes mismatch analysis only on valid reads
      mutations = find_mutations(validating["sequence"], reference) if valid else []
      #making an analysed read dict
      analyse_list.append({"read_id": read["read_id"],
                           "sample": read["sample"],
                           "group": read["group"],
                           "sequence": validating["sequence"],
                           "status": status,
                           "errors": validating["errors"],
                           "mutations": mutations})

    return analyse_list
````

````python
# Write your code here.
#helper function 1
def get_top_3_pos(position_counts, n = 3):
  """Getting top n (can be modified in the def line of code) positions sorted
     by descending order, with tie breaker by ascending order """
  #sorts mutation positions and finds the most frequent one
  sorted_pos = sorted(position_counts.keys(),
                      #lambda sorts positions in descending order, and in case of tie,
                      #the second pos will ensure the earlier position gets priority
                      key = lambda pos: (-position_counts[pos], pos))
  #get the top 3 mismatch position
  return sorted_pos[:n]

#helper function 2
def table_group_summ(group_name, group_samples):
  """Printing formatted summmary and substitution tables for each of the
  experimental groups"""
  print(f"Summary for group {group_name}")
  print("-" * 80)
  #table 1 of summarised results per sample
  print("[Table of Summarised Results]")
  header = (f"{'Sample':<12} | {'Valid':<7} | {'Invalid':<7} |"
   f"{'% w/ Mismatch':<15} | {'Total Events'} | {'Top Positions':<15}") #alligning the text to left within given charachter column
  print(header)
  print("-"* len(header))
  for sample_name, info in group_samples.items():
    pct = f"{info['mismatch_percentage']:.2f}%" if info["mismatch_percentage"] is not None else "0 valid" #rounding to 2 decimal places and addeing message if 0 valid reads
    pos_s = str(info["top_positions"])
    print(f"{sample_name:<12} | {info['valid_reads']:<7} | {info['invalid_reads']:<7} | "
            f"{pct:<15} | {info['mismatch_events']:<12} | {pos_s:<15}")
  #table 2 of substitution mutations, their count, percentage per 100 valid reads
  print("[Table of Substitution Mutations]")
  all_subs = sorted({sub for info in group_samples.values() for sub in info["substitution_counts"]})
  if not all_subs:
    print("No substitutions in this group")
    return
  sub_header = f"{'Mutation':<10}" + "".join(f" | {s:<16}" for s in group_samples)
  print(sub_header)
  print("-" * len(sub_header))
  for sub in all_subs:
    row = f"{sub:<10}"
    for sample_name, info in group_samples.items():
      count = info["substitution_counts"].get(sub, 0)
      if info["valid_reads"] > 0:
        rate = (count / info["valid_reads"]) * 100.0 #normalising to 100 reads
        cell = f"{count} ({rate:.1f}%)"
      else:
        cell = f"{count} (N/A)"
      row += f" | {cell:<16}"
    print(row)
  print("\n")

#main function
def summarise_samples(analysed_reads):
  """
 Combining read valitation and mismatch checks with subsequent calculations
 per sample per group

Parameters
----------
analysed reads : list of dict
    list ptoduced by Step 3, the keys for dict are specifed there too.

Returns
-------
dict
  dictionary keyed by sample name, containgin read counts, mismatch perc,
  substitution counts, position count and top 3 positions.

  returns {} if analysed_reads are empty.
  """
  #making sure this function works with required variable
  if not analysed_reads:
    return {}
  #new dictionary that collects totals for each sample
  raw_stats = {}
  #making another new dict every time a new sample name appears
  for read in analysed_reads:
    sample = read["sample"]
    if sample not in raw_stats:
      raw_stats[sample] = {"group": read["group"],
                           "valid_reads": 0,
                           "invalid_reads": 0,
                           "mismatch_reads": 0,
                           "mismatch_events": 0,
                           "substitution_counts": {},
                           "position_counts": {}}

    entry = raw_stats[sample]
    #if the function sees invalid sequence, it will not check for mismatches
    if read["status"] == "invalid":
            entry["invalid_reads"] += 1
    else: #if the sequence is valid it will check for mismatches and calculate subsequent mutation position etc.
            entry["valid_reads"] += 1
            mutations = read["mutations"]
            if mutations:
                entry["mismatch_reads"] += 1
                entry["mismatch_events"] += len(mutations)
                for mut in mutations:
                    sub_key = f"{mut['reference_base']}>{mut['read_base']}"
                    pos = mut["position"]
                    entry["substitution_counts"][sub_key] = entry["substitution_counts"].get(sub_key, 0) + 1
                    entry["position_counts"][pos] = entry["position_counts"].get(pos, 0) + 1
  #making a new dict for summary
  summary = {}
  #calculates mismatch percentage only with valid reads
  for sample, data in raw_stats.items():
      valid_count = data["valid_reads"]
      pct = (data["mismatch_reads"] / valid_count) * 100.0 if valid_count > 0 else None #makes sure there is no ZeroDivisionError
      #makinga dict for each of the samples
      summary[sample] = {"group": data["group"],
                        "valid_reads": valid_count,
                        "invalid_reads": data["invalid_reads"],
                        "mismatch_reads": data["mismatch_reads"],
                        "mismatch_percentage": pct,
                        "mismatch_events": data["mismatch_events"],
                        "substitution_counts": data["substitution_counts"],
                        "position_counts": data["position_counts"],
                        "top_positions": get_top_3_pos(data["position_counts"], n=3)} #gets top positions and adds it
  #separates into control and treatment groups
  groups = sorted({info["group"] for info in summary.values()})
  for group_name in groups:
      group_samples = {s: info for s, info in summary.items() if info["group"] == group_name}
      table_group_summ(group_name, group_samples)#calls each of the groups separetely
  return summary
````

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

Return a list with one dictionary per selected position:

    }

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.
def compare_positions(sample_summary, control_group: str = "control",
                      treatment_group: str = "treatment"):
    """
    Identify top 3 mismatch positions across control and treated groups and run Welch's t-test.

    Parameters
    ----------
    sample_summary : dict
        dictionary returned by `summarise_samples`.
    control_group : str *assuming that there is control group
        control samples
    treatment_group : str *assuming that there is treatment group
        treatment samples

    Returns
    -------
    list of dict
        each dictionary per top position containing sample sizes, group mean
        mismatch percentages, percentage-point difference (treatment - control),
        t-statistic, and p-value
    """
    #making sure to use sample_summary
    if not sample_summary:
        return []
    #combining mismatch counts across control and treatment group samples
    #making a variable to identify the highest mismatch position among all reads
    max_pos = 0
    for data in sample_summary.values():
        if data["group"] in (control_group, treatment_group) and data["position_counts"]:
            max_pos = max(max_pos, max(data["position_counts"].keys()))
    #ensuring that there is indeed some mismatches occured at least once
    if max_pos == 0:
        return []
    #making a 1d numpy int array to loop through every sample in each group
    combined_array = np.zeros(max_pos)
    for data in sample_summary.values():
        if data["group"] in (control_group, treatment_group):
            for pos, count in data["position_counts"].items():
              #adds each mismatch's position into array
                combined_array[pos - 1] += count #converts to 0-based indexing
    #sends made array to dictionary
    combined_counts = {pos + 1: int(count)#shifts back to 1-based indexing
                      for pos, count in enumerate(combined_array)
                      if count > 0}

    #using step 4 helper 1 to select top 3 positions
    top_positions = get_top_3_pos(combined_counts, n=3)

    #filtering out samples with 0 valid reads and print a report about excluding that sample
    usable_samples = {}
    for sample, data in sample_summary.items():
        if data["group"] in (control_group, treatment_group):
            if data["valid_reads"] == 0:
                print(f"Excluded sample '{sample}' ({data['group']}): 0 valid reads.")
            else:
                usable_samples[sample] = data
    #creating new list to store stat analysis (stores in dict) for each of top 3 mismatches
    results = []
    print("[ Statistical Analysis of Top 3 Mismatch Positions (Two-Sided Welch's t-test) ]")
    #loops over top 3 positions in test both groups
    for pos in top_positions:
        control_pcts = []
        treatment_pcts = []
        #calculating sample mismatch percentage
        for sample, data in usable_samples.items():
            mismatches_at_pos = data["position_counts"].get(pos, 0)
            pct = (mismatches_at_pos / data["valid_reads"]) * 100.0
            #sorting calculation results by testy groups
            if data["group"] == control_group:
                control_pcts.append(pct)
            elif data["group"] == treatment_group:
                treatment_pcts.append(pct)
        #couting mismatch percentage replicates in each test groups
        control_n = len(control_pcts)
        treatment_n = len(treatment_pcts)
        #calculating the mean and the difference in mismatch percentage for each group
        control_mean = float(np.mean(control_pcts)) if control_n > 0 else None
        treatment_mean = float(np.mean(treatment_pcts)) if treatment_n > 0 else None
        difference = (treatment_mean - control_mean) if (control_mean is not None and treatment_mean is not None) else None
        #running welch's t-test
        stats_res = calculate_statistics(control_pcts, treatment_pcts)
        #creating a dictionary to store positions and thei calculations
        entry = {
            "position": pos,
            "control_n": control_n,
            "treated_n": treatment_n,
            "control_mean": control_mean,
            "treated_mean": treatment_mean,
            "difference": difference,
            "statistic": stats_res["statistic"],
            "p_value": stats_res["p_value"],
        }
        results.append(entry)
        #formatting and printing summary line per position
        c_str = f"{control_mean:.2f}%" if control_mean is not None else "N/A"
        t_str = f"{treatment_mean:.2f}%" if treatment_mean is not None else "N/A"
        d_str = f"{difference:+.2f} pp" if difference is not None else "N/A"
        p_str = f"{stats_res['p_value']:.4e}" if stats_res["p_value"] is not None else "N/A"

        print(
            f"Position {pos}: {control_group} mean = {c_str} (n={control_n}), "
            f"{treatment_group} mean = {t_str} (n={treatment_n}), "
            f"diff = {d_str}, p = {p_str}"
        )

    return results
````
